##Exercise 1: scikit learn (sklearn)

Here is some synthetic data.

Run multiple regression and:
1. report which of the two predictors has the strongest relationship with the target variable.
2. which predictor is negatively related to the target variable?
3. report the proportion variance explained by the model.
4. what is the predicted value of y when both predictors are 0?


```python
# Import necessary libraries
import numpy as np
from sklearn.linear_model import LinearRegression

# Generate synthetic data
np.random.seed(42)
X = np.random.rand(100, 2)  # 100 samples, 2 predictors
y = 3 + 2.5 * X[:, 0] - 0.9 * X[:, 1] + np.random.normal(scale=0.4, size=100)  # Target with noise

# run multiple regression
model = LinearRegression()
model.fit(X, y)
print("coefficients: ", model.coef_)
print("Second predictor is negatively related to the target variable") if model.coef_[0] > model.coef_[1] else print("First predictor")

# report the proportion variance explained by the model
y_pred = model.predict(X)
print("Model variance: ", model.score(X, y))

# what is the predicted value of y when both predictors are 0?
print("intercept: ", model.intercept_)
```

##Exercise 2: statsmodels

Here is code to fit the same data with the statsmodels package.

Look at the model summary.

```python
# Import necessary libraries
import statsmodels.api as sm

# Add a constant term for the intercept
X_with_intercept = sm.add_constant(X)

# Fit the model
model_sm = sm.OLS(y, X_with_intercept).fit()

# Print the summary
print(model_sm.summary())
```
