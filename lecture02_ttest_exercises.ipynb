We will use the t-test functions from the scipy package. There is a different function for the three kinds of t-tests we covered:

2. independent samples t-test: [ttest_ind](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.ttest_ind.html)
3. paired samples t-test: [ttest_rel](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.ttest_rel.html)

The links above will lead you to the online help page for these functions.

Let's import the scipy stats package, and numpy

```python
import scipy.stats as stats
import numpy as np
```

>**Example:** Here is example code for how to do an independent samples t-test between two groups.

```python
groupA_data = np.array([4, 2, 5, 4, 5])
groupB_data = np.array([12, 8, 9, 8, 10])

t_statistic, p_value = stats.ttest_ind(groupA_data, groupB_data)

print("t-statistic:", t_statistic)
print("p-value:", p_value)
print("Significant:", p_value < 0.05)
```
